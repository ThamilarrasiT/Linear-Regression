3. Then we take the numerator and denominator for the divide

```python
x=[1,2,3,4,5]
y=[1,2,5,4,1]
n=len(x)
x_avg=sum(x)/n
y_avg=sum(y)/n
num=0
dem=0
for i in range(n):
  num+=(x[i]-x_avg)*(y[i]-y_avg)
  dem+=(x[i]-x_avg)**2
m=num/dem
c=y_avg-m*x_avg
loss=0
for i in range(n):
  y_pred=m*x[i]+c
  loss+=(y[i]-y_pred)**2
loss=loss/(2*n)
print("M=",m)
print("C=",c)
print("Loss=",loss)
```

```python
x=[1,2,3,4,5]
y=[1,3,2,4,5]
n=len(x)
x_avg=sum(x)/n
y_avg=sum(y)/n
num=0
dem=0
for i in range(n):
  num+=(x[i]-x_avg)*(y[i]-y_avg)
  dem+=(x[i]-x_avg)**2
m=num/dem
c=y_avg-m*x_avg


alpha=0.01
epochs=100
for epoch in range(epochs):
  dm=0
  dc=0
  for i in range(n):
      y_pred=m*x[i]+c
      error=y_pred-y[i]
      dm+=error*x[i]
      dc+=error
  dm=dm/n
  dc=dc/n
  m=m-alpha*dm
  c=c-alpha*dc
  loss=0
  for i in range(n):
    y_pred=m*x[i]+c
    loss+=(y[i]-y_pred)**2
  loss=loss/(2*n)
  print("Epoch=",epoch)
print("M=",m)
print("C=",c)
print("Loss=",loss)
print("dm=",dm)
print("dc=",dc)
```
